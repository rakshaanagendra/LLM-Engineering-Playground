```python
from transformers import AutoTokenizer

bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
gpt_tokenizer = AutoTokenizer.from_pretrained("gpt2")
t5_tokenizer = AutoTokenizer.from_pretrained("t5-small")

text = "Raksha is learning tokenization deeply"

print("BERT:", bert_tokenizer.tokenize(text))
print("GPT :", gpt_tokenizer.tokenize(text))
print("T5  :", t5_tokenizer.tokenize(text))
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")


texts = [
    "I love AI",
    "I love Stuttgart and machine learning"
]

encoded = tokenizer(
    texts,
    padding=True,
    truncation=True,
    return_tensors="pt"
)

print(encoded)
```

```python
print(tokenizer.convert_ids_to_tokens(encoded["input_ids"][0]))
```
