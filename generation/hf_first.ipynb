```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

result = generator("The future of AI is", 
        max_length=30,
        num_return_sequences=2,
        do_sample=True,
        temperature=0.8,
        top_k=5,
        top_p=0.9,
        repetition_penalty=1.2,
        )

print(result)
```
